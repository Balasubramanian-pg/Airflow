# Apache Airflow TaskFlow API: @dag Decorator

## Overview

The TaskFlow API transforms DAG definition using Python decorators instead of traditional operators. The `@dag` decorator converts a Python function into an Airflow DAG object.

### Key Benefits
- Reduced boilerplate code
- Python-native syntax
- Automatic XCom handling
- Improved readability

## 1. Basic @dag Implementation

```python
from airflow.decorators import dag
from datetime import datetime

@dag(
    dag_id='basic_taskflow_dag',
    start_date=datetime(2026, 9, 1),
    schedule_interval='@daily',
    catchup=False,
    tags=['example', 'taskflow']
)
def basic_dag():
    """Simple DAG demonstrating @dag decorator."""
    pass

dag_instance = basic_dag()
```

## 3. Dynamic DAG Generation

```python
from airflow.decorators import dag
from datetime import datetime

dag_configs = [
    {'name': 'sales', 'schedule': '0 7 * * *'},
    {'name': 'inventory', 'schedule': '0 8 * * *'},
    {'name': 'customer', 'schedule': '0 9 * * *'}
]

for config in dag_configs:
    @dag(
        dag_id=f'dynamic_{config["name"]}_dag',
        start_date=datetime(2026, 9, 1),
        schedule_interval=config['schedule'],
        catchup=False,
        tags=['dynamic', config['name']]
    )
    def dynamic_dag():
        pass
    
    globals()[f'{config["name"]}_dag'] = dynamic_dag()

print(f'Generated {len(dag_configs)} DAGs')
```

## 4. Context Manager Pattern with Tasks

```python
from airflow.decorators import dag, task
from datetime import datetime

@dag(
    dag_id='context_manager_example',
    start_date=datetime(2026, 9, 1),
    schedule_interval='@daily',
    catchup=False
)
def context_dag():
    """Tasks inside automatically belong to this DAG."""
    
    @task
    def extract():
        return {'records': 1000}
    
    @task
    def transform(data):
        print(f'Processing {data["records"]} records')
        return {'status': 'done'}
    
    @task
    def load(result):
        print(f'Loading: {result}')
        return True
    
    data = extract()
    result = transform(data)
    load(result)

dag_instance = context_dag()
```

## 5. DAG Parameters

```python
from airflow.decorators import dag, task
from datetime import datetime

@dag(
    dag_id='params_example',
    start_date=datetime(2026, 9, 1),
    schedule_interval='@daily',
    catchup=False,
    params={
        'environment': 'production',
        'batch_size': 500,
        'enable_notifications': True
    }
)
def params_dag():
    """Access parameters within tasks."""
    
    @task
    def process(**context):
        params = context['params']
        env = params.get('environment', 'dev')
        batch = params.get('batch_size', 100)
        print(f'Env: {env}, Batch: {batch}')
        return {'env': env, 'batch': batch}
    
    process()

dag_instance = params_dag()
```

## 6. Schedule Interval Formats

```python
from airflow.decorators import dag
from datetime import datetime, timedelta

# Cron expression
@dag(
    dag_id='cron_schedule',
    start_date=datetime(2026, 9, 1),
    schedule_interval='0 6 * * 1-5',
    catchup=False
)
def cron_dag():
    pass

# Preset
@dag(
    dag_id='preset_schedule',
    start_date=datetime(2026, 9, 1),
    schedule_interval='@hourly',
    catchup=False
)
def preset_dag():
    pass

# Timedelta
@dag(
    dag_id='timedelta_schedule',
    start_date=datetime(2026, 9, 1),
    schedule_interval=timedelta(hours=2),
    catchup=False
)
def timedelta_dag():
    pass

# Manual only
@dag(
    dag_id='manual_schedule',
    start_date=datetime(2026, 9, 1),
    schedule_interval=None,
    catchup=False
)
def manual_dag():
    pass

cron_dag()
preset_dag()
timedelta_dag()
manual_dag()
print('Created 4 DAGs with different schedules')
```

## 8. Common Pitfalls

```python
from airflow.decorators import dag
from datetime import datetime

# PITFALL 1: Forgetting instantiation
@dag(
    dag_id='needs_instantiation',
    start_date=datetime(2026, 9, 1),
    schedule_interval='@daily',
    catchup=False
)
def my_dag():
    pass

# CORRECT: Always instantiate
dag_instance = my_dag()

# PITFALL 2: Mutable defaults (avoid this)
# def bad_dag(config={}):  # Don't do this!

# PITFALL 3: Not setting catchup
@dag(
    dag_id='catchup_example',
    start_date=datetime(2026, 1, 1),
    schedule_interval='@daily',
    catchup=False  # Prevents backfilling
)
def no_catchup_dag():
    pass

no_catchup_dag()
print('Avoided common pitfalls')
```

## Summary

### Essential Parameters

| Parameter | Required | Purpose |
|-----------|----------|---------|
| `dag_id` | Yes | Unique identifier |
| `start_date` | Yes | Activation date |
| `schedule_interval` | No* | Run frequency |
| `catchup` | No | Backfill behavior |
| `default_args` | No | Task defaults |
| `tags` | No | Organization |

*Required unless manual triggering

### Checklist

✓ Instantiate DAG with `dag_instance = your_dag()`  
✓ Set `catchup=False` by default  
✓ Add meaningful `description` and `doc_md`  
✓ Use descriptive `tags`  
✓ Configure timeouts and retries  
✓ Limit `max_active_runs` for heavy DAGs
